```python
data="/content/dataset/data.yaml"
```

```python
from ultralytics import YOLO

# Load a model
model = YOLO('yolo26s.yaml')  # build a new model from scratch
model = YOLO('yolo26s.pt')  # load a pretrained model (recommended for training)


```

```python
# Use the model
results = model.train(data=data, epochs=30, imgsz=320, freeze=15, optimizer="SGD", batch=16, cls=0.3)  # train the model
```

```python
results = model.val()  # evaluate model performance on the validation set
```
